```python
from synthetic_data import generate_dataset
import time
import operator
import random


class Voter:
    def __init__(self, voter_id, age, county, party, income_bracket, turnout_history):
        self.voter_id = voter_id
        self.age = age
        self.county = county
        self.party = party
        self.income_bracket = income_bracket
        self.turnout_history = turnout_history

    def turnout_rate(self):
        if len(self.turnout_history) == 0:
            return 0.0
        return sum(self.turnout_history) / len(self.turnout_history)

    def turnout_year(self, year):
        year_map = {2016: 0, 2018: 1, 2020: 2, 2022: 3}
        return self.turnout_history[year_map[year]]

    def to_dict(self):
        return {
            "voter_id": self.voter_id,
            "age": self.age,
            "county": self.county,
            "party": self.party,
            "income_bracket": self.income_bracket,
            "t_2016": self.turnout_history[0],
            "t_2018": self.turnout_history[1],
            "t_2020": self.turnout_history[2],
            "t_2022": self.turnout_history[3],
            "turnout_rate": self.turnout_rate()
        }

    def __repr__(self):
        return (
            f"Voter(id={self.voter_id}, age={self.age}, county='{self.county}', "
            f"party='{self.party}', income='{self.income_bracket}', "
            f"turnout={self.turnout_history})"
        )


class Node:
    def __init__(self, voter):
        self.voter = voter
        self.next = None
```

```python
class VoterRegistry:

    def __init__(self, data=None):
        self.head = None
        self.tail = None
        self.size = 0 

        if data is not None:
            self.add_from_dataset(data)

    def __len__(self):
        return self.size

    def __iter__(self):
        current = self.head
        while current is not None:
            yield current.voter
            current = current.next

    def _record_to_voter(self, record):
        turnout_history = [
            record["t_2016"],
            record["t_2018"],
            record["t_2020"],
            record["t_2022"]
        ]
        return Voter(
            voter_id=record["voter_id"],
            age=record["age"],
            county=record["county"],
            party=record["party"],
            income_bracket=record["income_bracket"],
            turnout_history=turnout_history
        )

    def add_voter(self, voter):
        new_node = Node(voter)

        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:  # add to the tail instead of to the head
            self.tail.next = new_node
            self.tail = new_node

        self.size += 1

    def add_record(self, record): # convert to Voter, then add
        voter = self._record_to_voter(record)
        self.add_voter(voter)

    def add_from_dataset(self, data): # batch adding from dataset
        for record in data:
            self.add_record(record)

    def to_list(self): # turn voter to dictionary
        return [voter.to_dict() for voter in self]

    def show(self, n=None):
        rows = self.to_list()
        if n is not None:
            rows = rows[:n]
        for i, row in enumerate(rows):
            print(i, row)

    def _get_value(self, voter, by):   # return node data by all the variables
        if by == "voter_id":
            return voter.voter_id
        if by == "age":
            return voter.age
        if by == "county":
            return voter.county
        if by == "party":
            return voter.party
        if by == "income_bracket":
            return voter.income_bracket
        if by == "t_2016":
            return voter.turnout_history[0]
        if by == "t_2018":
            return voter.turnout_history[1]
        if by == "t_2020":
            return voter.turnout_history[2]
        if by == "t_2022":
            return voter.turnout_history[3]
        if by == "turnout_rate":
            return voter.turnout_rate()
        raise ValueError(f"Unsupported sort/search field: {by}")
    
    def sort(self, by):

        # insertion sort on linked list

        if self.head is None or self.head.next is None:
            return

        sorted_head = None
        current = self.head

        while current is not None:
            next_unsorted = current.next

            if sorted_head is None or self._get_value(current.voter, by) < self._get_value(sorted_head.voter, by):
                current.next = sorted_head
                sorted_head = current
            else:
                search = sorted_head
                while (
                    search.next is not None and
                    self._get_value(search.next.voter, by) <= self._get_value(current.voter, by)
                ):
                    search = search.next

                current.next = search.next
                search.next = current

            current = next_unsorted

        self.head = sorted_head

        # rebuild tail
        tail_finder = self.head
        while tail_finder.next is not None:
            tail_finder = tail_finder.next
        self.tail = tail_finder

    def search(self, field, op_string, value):
        # define operators
        ops = {
            "==": operator.eq,
            "!=": operator.ne,
            ">": operator.gt,
            ">=": operator.ge,
            "<": operator.lt,
            "<=": operator.le
        }

        if op_string not in ops:
            raise ValueError("Operator must be one of ==, !=, >, >=, <, <=")

        op_func = ops[op_string]
        matches = []

        index = 0
        current = self.head
        while current is not None:
            current_value = self._get_value(current.voter, field)
            if op_func(current_value, value):
                matches.append(index)
            current = current.next
            index += 1

        return matches


```

```python
def make_almost_sorted_registry(n, by="age", swaps=10):
    data = generate_dataset(n=n)
    registry = VoterRegistry(data)
    registry.sort(by) 
    # create a sorted list, then randomly swap some rows for benchmark test on almost sorted registry
    rows = registry.to_list() 
    for _ in range(swaps):
        i = random.randint(0, len(rows) - 1)
        j = random.randint(0, len(rows) - 1)
        rows[i], rows[j] = rows[j], rows[i]

    new_registry = VoterRegistry(rows)
    return new_registry


def make_reverse_sorted_registry(n, by="age"):
    data = generate_dataset(n=n)
    rows = sorted(
        data,
        key=lambda x: (
            x["age"] if by == "age" else x[by]
        ),
        reverse=True
    )  
    return VoterRegistry(rows) # create a reverse sorted registry for benchmark test


def benchmark_sort(n_values, by="age"): # test 
    results = []

    for n in n_values:
        random_registry = VoterRegistry(generate_dataset(n=n))
        t0 = time.perf_counter()
        random_registry.sort(by)
        t1 = time.perf_counter()

        almost_registry = make_almost_sorted_registry(n=n, by=by, swaps=max(1, n // 100))
        t2 = time.perf_counter()
        almost_registry.sort(by)
        t3 = time.perf_counter()

        reverse_registry = make_reverse_sorted_registry(n=n, by=by)
        t4 = time.perf_counter()
        reverse_registry.sort(by)
        t5 = time.perf_counter()

        results.append({
            "n": n,
            "random_sort_seconds": t1 - t0,
            "almost_sorted_seconds": t3 - t2,
            "reverse_sorted_seconds": t5 - t4
        })

    return results


def benchmark_search(n_values):
    results = []

    for n in n_values:
        registry = VoterRegistry(generate_dataset(n=n))

        t0 = time.perf_counter()
        registry.search("age", ">", 30)
        t1 = time.perf_counter()

        t2 = time.perf_counter()
        registry.search("county", "==", "Fulton")
        t3 = time.perf_counter()

        results.append({
            "n": n,
            "search_age_gt_30_seconds": t1 - t0,
            "search_county_fulton_seconds": t3 - t2,
        })

    return results
```
